```python
import random

my = int(input("請輸入 0.剪刀 1.石頭 2.布:"))
com = random.randint(0, 2)

symbols = ["剪刀", "石頭", "布"]
print("我出的:", symbols[my])
print("電腦的:", symbols[com])

# 我出的是電腦的下一個拳: win
if my == (com + 1) % 3:
    print("win")
elif com == (my + 1) % 3:
    print("lose")
else:
    print("even")
```

- 取商: 分組
    - 0 // 3 -> 0
    - 1 // 3 -> 0
    - 2 // 3 -> 0
    - 3 // 3 -> 1
    - 4 // 3 -> 1
    - 5 // 3 -> 1
    - 6 // 3 -> 2

- 縮排快捷
    - TAB
    - Shift+TAB

```python
# 練習題
import random

# my = int(input("0-5"))
# com = random.randint(0, 5)
my = 3
com = 5

symbols = ["剪刀(0)", "石頭(0)", "布(0)",
           "剪刀(1)", "石頭(1)", "布(1)"]
print("我出的:", symbols[my])
print("電腦的:", symbols[com])

# 先確認是否同組
my_g = my // 3
com_g = com // 3
if my_g == com_g:
    print("同組")
    # case1. my, com -> 0/1/2
    # case2. my, com -> 3/4/5 - 3 -> 0/1/2
    my2, com2 = my % 3, com % 3
    # 我出的是電腦的下一個拳: win
    if my2 == (com2 + 1) % 3:
        print("win")
    elif com2 == (my2 + 1) % 3:
        print("lose")
    else:
        print("even")
else:
    print("不同組")
    print("even")
```
