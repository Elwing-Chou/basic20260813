```python
a, b = 30, 7
result = str(a // b) + "."
a = a % b

for i in range(10):
    # 2 -> 20
    a10 = a * 10
    # 20 // 7 -> 2 -> 0.2
    result = result + str(a10 // b)
    print(result)
    # 20 % 7 -> 6
    a = a10 % b
    if a == 0:
        print("整除")
        break
```

```python
import random

total = 100000000
count = 0
for i in range(total):
    x = random.uniform(-1, 1)
    y = random.uniform(-1, 1)
    if x ** 2 + y ** 2 <= 1:
        count = count + 1

ratio = count / total
reference_area = 4
true_area = ratio * reference_area
print("丟到裡面的數目:", count)
print("面積:", true_area)
```

```python
n = 2000000
radius = 1
# 0.25
interval = radius / n
total_area = 0
for i in range(n):
    # i = 0 1 2 3
    # x = 0.25 0.5 0.75 1.0
    x = i * interval + interval
    y = (radius ** 2 - x ** 2) ** 0.5
    # 單一長方形的面積
    area = interval * y
    total_area = total_area + area

print("面積:", 4 * total_area)
```
