```python
import random

win, even, lose = 0, 0, 0
# cond1: win < 3
# cond2: lose < 3
while (win < 3) and (lose < 3):
    # -1: lose 0: even 1: win
    result = random.randint(-1, 1)
    if result == -1:
        lose = lose + 1
    elif result == 0:
        even = even + 1
    elif result == 1:
        win = win + 1
    print("w:e:l", win, even, lose)

if win > lose:
    print("WIN")
else:
    print("LOSE")
```
