```python
s = "The lord of the rings-The fellowship of the ring"
# s = "fox"
secret = 3
result = ""

for c in s:
    c_int = ord(c)
    if ord("A") <= c_int <= ord("Z"):
        n = c_int - ord("A")
        n2 = (n + secret) % 26 + ord("A")
        result = result + chr(n2)
    elif ord("a") <= c_int <= ord("z"):
        n = c_int - ord("a")
        n2 = (n + secret) % 26 + ord("a")
        result = result + chr(n2)
    else:
        result = result + c

f = open("a.txt", "w", encoding="utf-8")
f.write(result)
f.close()
```
