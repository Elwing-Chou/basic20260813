- 如果...否則如果...否則

```python
# bmi->體重(kg)/身高(m)^2
weight = float(input("請輸入體重:"))
height = float(input("請輸入身高:"))
bmi = weight / (height / 100) ** 2

print("你的體重是:" + str(weight))
print("你的身高是:" + str(height))
print("計算出來的bmi:" + str(bmi))

if bmi > 25:
    print("過重")
elif bmi > 18:
    print("正常")
else:
    print("過輕")
print("!!!!")
```
