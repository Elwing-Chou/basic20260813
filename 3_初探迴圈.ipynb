```python
layer = 10
lsymbol, rsymbol = " ", "o"
i = 0
while i < layer:
    lcount = layer - i - 1
    rcount = i + 1
    line = lsymbol * lcount + rsymbol * rcount
    print(line)
    i = i + 1
```

```python
layer = 10
lsymbol, rsymbol = " ", "o"
i = 0
while i < layer:
    lcount = i
    rcount = layer - i
    line = lsymbol * lcount + rsymbol * rcount
    print(line)
    i = i + 1
```

```python
layer = 10
s1, s2, s3 = " ", "o", " "
i = 0
while i < layer:
    c1 = i
    c2 = (2 * layer - 1) - 2 * i
    c3 = i
    line = s1 * c1 + s2 * c2 + s3 * c3
    print(line)
    i = i + 1
```

```python
layer = 10
i = 0
while i < layer:
    if i % 2 == 0:
        s1, s2, s3 = " ", "o", " "
    else:
        s1, s2, s3 = " ", "x", " "

    c1 = i
    c2 = (2 * layer - 1) - 2 * i
    c3 = i
    line = s1 * c1 + s2 * c2 + s3 * c3
    print(line)
    i = i + 1
```

```python
layer = 10
s1_list = [" "]
s2_list = ["o", "x", "u"]
s3_list = [" "]


i = 0
while i < layer:
    # 拿取我的symbol 0 1 2 0
    s1 = s1_list[i%len(s1_list)]
    s2 = s2_list[i%len(s2_list)]
    s3 = s3_list[i%len(s3_list)]
    # 算出各自symbol的數目
    c1 = i
    c2 = (2 * layer - 1) - 2 * i
    c3 = i
    line = s1 * c1 + s2 * c2 + s3 * c3
    print(line)
    i = i + 1
```
