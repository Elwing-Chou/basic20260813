```python
import pandas as pd
import json
import os

d = input("日期(yyyy/mm/dd):")
category = input("物品:")
price = input("價格:")

# 如果這個檔案不存在
fn = "記帳3.csv"
if not os.path.exists(fn):
    print("檔案不存在, 創建新的[]")
    table = []
else:
    print("檔案存在, 讀取裡面的內容來變成[]")
    df = pd.read_csv(fn, encoding="utf-8")
    data_str = df.to_json(force_ascii=False, orient="records")
    table = json.loads(data_str)

row = {
    "日期":d,
    "物品":category,
    "價格":price
}
table.append(row)
df = pd.DataFrame(table)
df.to_csv(fn, encoding="utf-8", index=False)
```
