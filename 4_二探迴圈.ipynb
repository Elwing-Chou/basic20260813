```python
import ColabTurtlePlus.Turtle as turtle
import random

turtle.clearscreen()
turtle.setup(500, 500)
turtle.showborder()
turtle.shape("turtle")
turtle.pensize(2)
turtle.speed(10)

n = 30
distance = 100
angle_1 = 360 / n
angle_2 = 180 - angle_1
for i in range(n):
    # 開始填色
    turtle.begin_fill()
    r = random.randint(0, 255)
    g = random.randint(0, 255)
    b = random.randint(0, 255)
    turtle.fillcolor(r, g, b)

    turtle.forward(distance)
    turtle.right(angle_1)
    turtle.forward(distance)
    turtle.right(angle_2)
    turtle.forward(distance)
    turtle.right(angle_1)
    turtle.forward(distance)
    turtle.right(180)
    # 結束填色
    turtle.end_fill()
```
